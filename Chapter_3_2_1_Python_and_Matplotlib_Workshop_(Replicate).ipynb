```python
import numpy as np
import matplotlib.pyplot as plt

# Definir la función
x = np.linspace(-10, 10, 1000)  # Ampliar el rango de x
y = 0.25 * (x + 4) * (x + 2) * (x - 3)

# Crear la figura con un tamaño reducido
plt.figure(figsize=(8, 4))  # Reducir el tamaño del gráfico

# Graficar la función
plt.plot(x, y, linewidth=2, label="Function: y = 0.25 *(x+4)* (x+2) * (x-3)")

# Ajustar los ejes
plt.xlim(-10, 10)  # Ampliar el eje X
plt.ylim(-20, 35)  # Mantener el mismo eje Y

# Dibujar los ejes en verde punteado
plt.axhline(0, color='green', linestyle='--', linewidth=1.5, label="Axis")
plt.axvline(0, color='green', linestyle='--', linewidth=1.5)

# Calcular las raíces de la función
roots = [-4, -2, 3]
y_roots = [0, 0, 0]

# Graficar las raíces como estrellas rojas
plt.scatter(roots, y_roots, color='red', s=70, marker='*', label="Roots")

# Anotar las raíces con flechas en diferentes posiciones
annotations = [
    (-40, 50),  # Desplazamiento para root 1
    (0, 40),    # Desplazamiento para root 2
    (50, -40)   # Desplazamiento para root 3
]

# Agregar anotaciones con flechas
for i, (root, offset) in enumerate(zip(roots, annotations)):
    plt.annotate(f'root {i+1}',
                 (root, 0),
                 textcoords="offset points",
                 xytext=offset,
                 ha='center',
                 fontsize=10,
                 fontweight='bold',
                 arrowprops=dict(facecolor='black', arrowstyle='->', lw=2))

# Agregar la leyenda fuera del gráfico
plt.legend(loc="upper left", bbox_to_anchor=(1.05, 1), borderaxespad=0.)

# Ajustar los márgenes para que la leyenda no se corte
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```
